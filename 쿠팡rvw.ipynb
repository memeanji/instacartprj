```python
import pandas as pd
from collections import Counter
from konlpy.tag import Okt

# Okt 형태소 분석기 객체 생성
okt = Okt()

# 리뷰 데이터 불러오기
df = pd.read_csv("coupang_rvw/df_all_LLM.csv")

# 1. 3점 이하 리뷰 필터링
low_score_reviews = df[df['유저별별점'] <= 3]['리뷰본문']

# 2. 단어별 빈도 집계 (명사 중심)
all_words = []

for text in low_score_reviews:
    if isinstance(text, str):
        # 명사만 추출
        nouns = okt.nouns(text)
        all_words.extend(nouns)

# 불용어 예시 (필요하면 추가/수정)
stopwords = ['구매', '배송', '상품', '리뷰', '주문', '그냥', '진짜', '이번',
              '정도', '부분', '하나']


# 불용어 제거
filtered_words = [word for word in all_words if word not in stopwords and len(word) > 1]  # 한 글자 제외

word_counts = Counter(filtered_words)

# 3. 빈도 상위 단어 출력
top_words = word_counts.most_common(20)
top_words

```

1. 명사 + 형용사 조합 추출

```python
from konlpy.tag import Okt
import pandas as pd

okt = Okt()

# 리뷰 데이터 불러오기
df = pd.read_csv("coupang_rvw/df_all_LLM.csv")
low_score_reviews = df[df['유저별별점'] <= 3]['리뷰본문']

# 명사+형용사 조합 빈도수 세기
combo_counter = Counter()

for text in low_score_reviews:
    if isinstance(text, str):
        pos_tags = okt.pos(text)
        # pos_tags는 [('단어', '품사'), ...] 리스트 형태
        for i in range(len(pos_tags) - 1):
            word1, tag1 = pos_tags[i]
            word2, tag2 = pos_tags[i + 1]
            if tag1 == 'Noun' and tag2 == 'Adjective':
                combo_counter[(word1, word2)] += 1

# 상위 20개 출력
top_combos = combo_counter.most_common(20)
top_combos
```

-> 이는 실제로 상품에 대한 과도한 긍정 평가가 대부분 5점 리뷰에 몰려 있었을 가능성을 시사합니다.

2. 부정 리뷰는 구성 비슷하나 비율 달라졌을 수 있음
- 여전히 부정적인 감정은 배송, 기타, 상품에 걸쳐 있음.

- 감성 분석 모델이 상품 관련 불만 표현에 더 민감하게 반응한 것으로 보입니다.

| 항목    | 5점 리뷰 포함   | 5점 리뷰 제외                 |
| ----- | ---------- | ------------------------ |
| 긍정 리뷰 | 배송 + 상품    | **배송 집중**                |
| 부정 리뷰 | 배송, 상품, 기타 | **거의 비슷하나 비율 변화** 가능성 있음 |

- count_table['긍정'] : 특정 경험요소별 긍정 리뷰 개수

- count_table['부정'] : 특정 경험요소별 부정 리뷰 개수

- (count_table['긍정'] + count_table['부정']) : 해당 경험요소에 대해 긍정과 부정 리뷰의 총합 (중립 제외)

- => 긍정비율 = 긍정 리뷰 개수 ÷ (긍정 리뷰 개수 + 부정 리뷰 개수)
- => 부정비율 = 부정 리뷰 개수 ÷ (긍정 리뷰 개수 + 부정 리뷰 개수)

- => 긍정비율과 부정리뷰 합하면 1이 될 수 있게

```python
import pandas as pd
from collections import Counter
from konlpy.tag import Okt

# 데이터 로드
df = pd.read_csv(r"C:\Users\User\Desktop\프로젝트\python_prj\coupang_rvw\LLM_Coupang 감성분석_결과1to4.csv")

# Okt 형태소 분석기
okt = Okt()

# 감성점수 컬럼명 예시: '감성점수' (1: 긍정, 0: 중립, -1: 부정)

# 1) 신선/비신선 분리
fresh_df = df [df ['라벨'] == '신선']
nonfresh_df = df[df['라벨'] == '비신선']

# 2) 긍정(1) / 부정(-1) 리뷰만 추출 (중립(0) 제외)
fresh_df = fresh_df[fresh_df['감성점수'].isin([1, -1])]
nonfresh_df = nonfresh_df[nonfresh_df['감성점수'].isin([1, -1])]

# 3) 감성점수 숫자 -> 문자열 라벨 매핑 함수
def score_to_label(score):
    return '긍정' if score == 1 else '부정'

# 4) 명사 추출 함수 및 단어 추출 함수 (기존과 동일)
from collections import Counter
from konlpy.tag import Okt
okt = Okt()

def extract_service_words(df, stopwords, service_keywords):
    texts = df['리뷰본문'].dropna()
    words = []
    for text in texts:
        nouns = [w for w in okt.nouns(text) if w not in stopwords]
        filtered = [w for w in nouns if w in service_keywords]
        words.extend(filtered)
    return Counter(words)

stopwords = ['것', '때', '수', '개', '더', '좀', '이런', '저런', '그냥', '정말', '거', '진짜', '기분', '생각', '자체', '정도']
service_keywords = ['배송', '포장', '상태', '파손', '시간', '지연', '기사', '응대', 'CS', '교환', '환불', '문의', '전화', '상담', '친절', '불친절']

# 5) 감성점수별로 필터링 후 단어 빈도 추출

# 신선 긍정
fresh_pos = extract_service_words(fresh_df[fresh_df['감성점수'] == 1], stopwords, service_keywords)
# 신선 부정
fresh_neg = extract_service_words(fresh_df[fresh_df['감성점수'] == -1], stopwords, service_keywords)
# 비신선 긍정
nonfresh_pos = extract_service_words(nonfresh_df[nonfresh_df['감성점수'] == 1], stopwords, service_keywords)
# 비신선 부정
nonfresh_neg = extract_service_words(nonfresh_df[nonfresh_df['감성점수'] == -1], stopwords, service_keywords)

# 6) 출력용 DataFrame 만드는 함수
def get_top_keywords_df(counter_obj, label, top_n=10):
    top_items = counter_obj.most_common(top_n)
    df = pd.DataFrame(top_items, columns=[label + ' 키워드', label + ' 빈도'])
    return df

fresh_pos_df = get_top_keywords_df(fresh_pos, '신선 긍정')
fresh_neg_df = get_top_keywords_df(fresh_neg, '신선 부정')
nonfresh_pos_df = get_top_keywords_df(nonfresh_pos, '비신선 긍정')
nonfresh_neg_df = get_top_keywords_df(nonfresh_neg, '비신선 부정')

# 7) 결과 출력 (라벨은 문자열로, 데이터 내부는 점수로 유지됨)
print(fresh_pos_df)
print(fresh_neg_df)
print(nonfresh_pos_df)
print(nonfresh_neg_df)
```

```python
from collections import Counter
from konlpy.tag import Okt
import pandas as pd

okt = Okt()

stopwords = ['것', '때', '수', '개', '더', '좀', '이런', '저런', '그냥', '정말', '거', '진짜', '기분', '생각', '자체', '정도']
service_keywords = ['배송', '포장', '상태', '파손', '시간', '지연', '기사', '응대', 'CS', '교환', '환불', '문의', '전화', '상담', '친절', '불친절']

def get_word_counts(df, label):
    texts = df[df['감성분류_가중치'] == label]['리뷰본문'].dropna()
    words = []
    for text in texts:
        nouns = [w for w in okt.nouns(text) if w not in stopwords and w in service_keywords]
        words.extend(nouns)
    counter = Counter(words)
    total_words = sum(counter.values())
    return counter, total_words

pos_counter, pos_total = get_word_counts(df, '긍정')
neg_counter, neg_total = get_word_counts(df, '부정')

all_keywords = set(pos_counter.keys()) | set(neg_counter.keys())

data = []
for kw in all_keywords:
    pos_count = pos_counter.get(kw, 0)
    neg_count = neg_counter.get(kw, 0)
    pos_freq = pos_count / pos_total if pos_total > 0 else 0
    neg_freq = neg_count / neg_total if neg_total > 0 else 0
    data.append([kw, pos_count, neg_count, pos_freq, neg_freq])

df_freq = pd.DataFrame(data, columns=['키워드', '긍정빈도', '부정빈도', '긍정비율', '부정비율'])

# 빈도 합으로 정렬 (긍정 + 부정 빈도)
df_freq['총빈도'] = df_freq['긍정빈도'] + df_freq['부정빈도']

# 빈도 순 내림차순 정렬 출력
df_freq.sort_values(by='총빈도', ascending=False)
```

### 그러면 내가 지금 가중치를 어떻게 주었냐면

| 단계 | 설명                      | 결과                        |
| -- | ----------------------- | ------------------------- |
| 1  | 리뷰에서 경험요소 추출            | `'배송'`, `'상품'` 등          |
| 2  | 감성 점수 부여                | `-1`, `0`, `+1` 등         |
| 3  | 경험요소별 평균 감성점수 → **가중치** | `'배송': 0.3`, `'상품': -0.2` |
| 4  | 감성점수 × 경험요소 가중치         | 감성 온도 반영된 점수              |
| 5  | 전체 평균                   | 브랜드/제품에 대한 종합적 감성 점수      |
